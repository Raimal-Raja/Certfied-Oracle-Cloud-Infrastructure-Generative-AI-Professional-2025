### OCI Generative AI service
Fully managed service that provides a set of customizable large language Model (LLMs) available via a single API to build generative AI applications.

### Choice of Model
high performing pretrained foundational models from meta and cohere.

### Flexible fine-tuning
create custom models by fine-tunning foundational models with your own data set

### Dedicated AI clusters
GPU based compute resources that host your fine-tuning and inference workloads.
