### OCI Generative AI service
Fully managed service that provides a set of customizable large language Model (LLMs) available via a single API to build generative AI applications.

### Choice of Model
high performing pretrained foundational models from meta and cohere.

### Flexible fine-tuning
create custom models by fine-tunning foundational models with your own data set

### Dedicated AI clusters
GPU based compute resources that host your fine-tuning and inference workloads.

### Chat Model Parameter
#### Maximum output tokens
Max number of tokens model generates per response.
#### Preamble Override
An initiial guideline message that can change the model's overall chat behavior and conversation style.

If specified, the model's default preamble is replaced with provided preamble.

#### temperature
controls the randomness of output. to generate the same output for a prompt every time you run that prompt use 0.

lower values are used in tasks with a 'correct' answer (Q&A). Higher values enable the model to generate more 'creative outputs but might generate hallucination
