## 🧱 Understanding Tokens

Before diving into chat models, let's first understand **tokens** — the fundamental building blocks that large language models (LLMs) process.

Large language models understand **tokens**, not characters.

- A **token** can be:
  - A part of a word,
  - A complete word, or
  - A punctuation mark.

### 🧩 Examples:
- The word **"apple"** → 1 token  
- The word **"friendship"** → 2 tokens → `friend` + `ship`

👉 On average:
- **Simple text:** ≈ 1 token per word  
- **Complex text:** ≈ 2–3 tokens per word  

### 🧮 Example Sentence

> "Many words map to one token, but some don't, indivisible."

A tokenizer would break it into **15 tokens**, though there are only **10 words**.  
Tokens can also include punctuation such as commas, apostrophes, or periods.

Some rare words (e.g., *indivisible*) split into multiple tokens (`indi` + `visible`).

The **OCI Generative AI service** provides several pretrained foundational chat models:

### 🧠 1. Command-R-Plus
- **Type:** Instruction-following conversational model  
- **Prompt length:** Up to **128,000 tokens**

- **Use cases:** Q&A, chatbots, information retrieval, and sentiment analysis

- **Type:** Smaller, faster version of Command-R-Plus  
- **Prompt length:** Up to **16,000 tokens**

- **Use cases:** Ideal when **speed and cost** are important  
- **Key trait:** Nearly as capable as Command-R-Plus, but more efficient

### 🦙 3. Llama 3.1 (by Meta)
- **Versions:** 70B and 400B parameter models  
- **Prompt/Response:** Up to **128,000 tokens** each  
- **400B model:** Largest publicly available model  
- **Use cases:** Complex, enterprise-grade applications

## ⚙️ Parameters That Control Chat Model Output

You can **tune model behavior** using various parameters:

| Parameter | Description |
|------------|--------------|
| **Max Output Tokens** | Controls how many tokens the model generates in its response |
| **Preamble Override** | Sets or overrides the initial context/tone of the chat |
| **Temperature** | Controls randomness of the model’s output |
| **Top-k** | Limits choices to the top *k* most probable tokens |
| **Top-p** | Chooses from tokens whose cumulative probability ≤ *p* |
| **Frequency Penalty** | Reduces repetition by penalizing frequently used tokens |
| **Presence Penalty** | Penalizes any token that has already appeared once |

---

## 🧾 Preamble Override

**Preamble** = The initial context or guiding message that shapes the model’s tone and behavior.

### ✏️ Example:
Default preamble:  
> "You are a helpful assistant."

Overridden preamble:  
> "You are a pirate. Answer every question in a pirate tone."

Result → The model’s responses change to match the *pirate style*.

---

## 🔥 Temperature — Controlling Randomness

The **temperature** parameter affects how creative or random the model’s responses are.

### Example Prompt:
> "The sky is"

The model chooses the next token based on **probability**:

| Token | Probability |
|--------|--------------|
| blue | 0.80 |
| limit | 0.15 |
| red | 0.05 |

### 🎛️ Temperature Settings:
- **0.0 → Deterministic:** Always picks *blue*  
- **1.0 → Creative:** May pick *limit* or *red*  
- **Higher value → More random**, less predictable results  

Thus, **temperature** controls how "surprising" or "diverse" the output can be.

---

## 🎯 Top-k Sampling

**Top-k** controls how many of the most probable tokens the model can choose from.

### Example:
> "The name of that country is the ..."

| Token | Probability |
|--------|--------------|
| United | 0.60 |
| Netherlands | 0.30 |
| Czech | 0.10 |
| Greece | 0.05 |
| India | 0.03 |

If **Top-k = 3**, the model can only choose among:
**United**, **Netherlands**, and **Czech**.

The rest are ignored.

---

## 🎲 Top-p Sampling (Nucleus Sampling)

**Top-p** selects tokens whose **cumulative probability** ≤ *p*.

### Example:
If **Top-p = 0.15 (15%)**  
→ Model picks the smallest set of tokens whose combined probabilities ≤ 15%.  
So it might include *United* and *Netherlands* but exclude *Czech* and others.

Increasing **p** allows for a wider, more diverse set of possible next tokens.

You can combine **Top-p** and **Top-k** for finer control.

---

## 🔁 Frequency and Presence Penalties

Both help avoid repetitive or unnatural responses.

### ⚙️ Frequency Penalty:
- Penalizes tokens that have appeared *many times* before.  
- Reduces the likelihood of overused words reappearing.

### ⚙️ Presence Penalty:
- Penalizes any token that has already appeared *at least once*.  
- Ensures more variation in responses.

📌 **Use Case:**  
If your model repeats phrases, slightly increasing these penalties encourages more natural variation.
