# 🧠 OCI Generative AI Service — Demonstration Overview

## 🌍 Region Selection

Welcome to this demo of the **OCI Generative AI Service**.  
In this demonstration, we begin by logging in to the **OCI Console**.  

Currently, we are in the **Germany Central (Frankfurt)** region.  
OCI Generative AI service is available in a few **select regions**, so ensure that the region you select supports this service.  
For this demo, the **Frankfurt region** is chosen.

---

## ⚙️ Accessing the Service

To access the service:

1. Click the **burger menu** (☰) on the left-hand side.
2. Under **Analytics and AI**, select **AI Services**.
3. Click **Generative AI**.

This opens the **Generative AI Dashboard**, which includes:

- **Service Tour** – a guided video overview.  
- **Documentation** – API references and usage guides.  
- **Playground** – a *visual interface* for exploring pretrained or custom models without coding.

---

## 🧩 Playground Overview

The **Playground** allows users to:

- Experiment with **hosted, pretrained, and custom models**.
- Test **prompts**, **parameters**, and **responses** visually.
- **View code snippets** in Python or Java to integrate models into your applications.

Once you’re satisfied with your experiment, you can copy the generated code and use it directly in your **IDE** or **Jupyter Notebook**.

---

## 💻 Dashboard Components

The dashboard includes:

- **Dedicated AI Clusters** → GPU-based compute resources for fine-tuning and hosting models.  
- **Custom Models** → User fine-tuned models.  
- **Endpoints** → URLs where models are hosted for inference.

---

## 💬 Using Chat Models

Inside the Playground, two model classes are visible:

- **Chat Models**
- **Embedding Models**

### Chat Models
When selecting **Chat Models**, available options include:

- `Command-R`
- `Command-R-Plus`
- `Meta Llama 3 (70B Instruct Model)`

Each model differs in **capability** and **token limits**:
| Model | Max Tokens | Notes |
|--------|-------------|-------|
| Command-R-Plus | 128,000 | Most powerful |
| Command-R | 16,000 | Balanced and affordable |
| Llama 3 | 8,000 | Lightweight option |

You can view details by selecting **Model Details**, which links to OCI documentation showing token limits and regional availability.

## 💡 Chat Example

Example prompt:
> *"Teach me how to fish."*

The model generates step-by-step guidance.  
Because it’s a **chat model**, it retains context.  
A follow-up question like:
> *"Describe step 3."*

will return a contextualized response based on the previous conversation.

Once satisfied with the result, click **View Code** to reveal the auto-generated API integration code.

## 🎭 Customizing Model Behavior

You can modify:

- **Preamble Override** → Changes the model’s tone or role (e.g., “You are a travel advisor who speaks like a pirate.”)
- **Temperature** → Adjusts output randomness.

Example:  
Changing the preamble to a pirate theme will make responses mimic pirate-style dialogue!

## 🔢 Embedding Models

### What Are Embeddings?
Embeddings convert **text into numerical vectors** representing **semantic meaning**.  
They enable **semantic search** — searching based on *meaning*, not just *keywords*.

### Example
Use the *HR Help Center* dataset with 41 articles.

When converted into embeddings and plotted, semantically similar articles (e.g., “Technical Skills”, “Career Development”, “Business Skills”) cluster together.

Even though plotted in **2D**, embeddings actually exist in **hundreds of dimensions** (e.g., 384+).

This visualization demonstrates that semantically similar texts have **numerically similar embeddings**.

---

## 📊 Applications of Embeddings

If a user searches for “skills,” the system retrieves clustered articles related to “technical skills” or “career development.”

This principle powers:
- **Semantic search**
- **Recommendation systems**
- **Contextual document retrieval**

As with chat models, you can view and copy **Python or Java code** for embeddings integration.

## 🧮 Dedicated AI Clusters

These clusters are **GPU-based compute resources** used for:
- Fine-tuning models  
- Serving inference traffic

To create a cluster:
1. Click **Create Dedicated AI Cluster**.
2. Choose a **name**.
3. Select **purpose**: fine-tuning or inference.
4. Choose **pretrained models**.
5. Click **Create**.

---

## 🧰 Custom (Fine-Tuned) Models

You can fine-tune models by:

1. Clicking **Create Model**.
2. Providing a **name**.
3. Selecting a **base model**.
4. Choosing a **fine-tuning method**.
5. Assigning a **dedicated AI cluster**.

Fine-tuning allows specialized model behavior for tasks like domain-specific Q&A, customer service, or custom text generation.

---

## 🌐 Creating Endpoints

Once a fine-tuned model is ready:

1. Click **Create Endpoint**.
2. Provide:
   - Endpoint name  
   - Hosted model details  
   - Cluster configuration  

This endpoint serves inference traffic, allowing your application to send queries and receive generated responses.
