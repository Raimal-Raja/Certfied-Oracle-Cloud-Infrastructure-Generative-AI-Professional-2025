## LangChain Components
langChain is a framework for developing applications powered by language models.

it offers a multitude of components that help us build components that help us build LLM-powered applications.

a few components that are used to build our chatbot.


LLMs in langChain refer to pure text completion models.

they take a string prompt as input and ouput a string completion


The core element of any language model application is the model

chat models are often  backed by LLM but are tuned specifically for having conversions. they take a list of chat messages as input and return as AI message as output
