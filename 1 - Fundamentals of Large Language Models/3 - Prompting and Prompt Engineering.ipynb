## Prompting Overview

Let’s revisit our earlier example:

> “I wrote to the zoo to send me a pet. They sent me a—”

As we know, an LLM computes a distribution over the next word in the sequence. The probabilities under each possible next word represent this distribution.

In this section, we’ll discuss how we can **influence** that distribution—how to update or affect the probabilities.

There are **two main ways**:
1. **Prompting**
2. **Training**

---

## Prompting: Changing the Input

Prompting refers to altering the **content or structure** of the input we pass to the model. Changing even one word—or even a space—can change the probability distribution over vocabulary words.

For example, if we append the word *“little”* to our sentence:
> “I wrote to the zoo to send me a little—”

The model shifts its probabilities—small animals like *dog* or *cat* become more likely, while larger animals like *lion* or *panther* become less likely.

This happens because during **pre-training**, the model learns patterns such as the frequency of bigrams like *“little dog”* or *“little cat.”*

---

## Prompt Engineering

**Prompt engineering** is the iterative process of refining a model’s input text to induce a desired output or probability distribution.

It can be:
- **Highly effective**, when done carefully.
- **Unintuitive**, because small input changes can have large, unpredictable effects.

For example, even rephrasing:
> “I wanted a pet, so I asked the zoo to send me one.”

...may alter the model’s output unpredictably.

Despite these challenges, prompt engineering has proven very valuable, and numerous academic and industrial strategies have emerged.

---

## In-Context Learning and k-Shot Prompting

**In-context learning** refers to including examples of the task within the prompt itself. Importantly, **no model parameters are changed**—the model simply uses the examples as context.

**k-shot prompting** means including *k* examples of the task before asking the model to perform it.

### Example (from the GPT-3 Paper)
Prompt:

```

Translate English to French:

sea otter => loutre de mer
cheese => fromage
seal => phoque
walrus =>

```

This is a **3-shot prompt**, showing three examples before asking the model to complete the fourth translation.

When no examples are given, we call it **zero-shot prompting**.

## Examples of Different Prompts

### 1. Two-Shot Addition Prompt
```

Q: 2 + 3 =
A: 5
Q: 4 + 6 =
A: 10
Q: 1 + 8 =
A:

```
The model generates a probability distribution for what comes after “A:”.

### 2. Instruction-Based Prompt (MPT-Instruct)
```

You are an AI model that follows these rules: be concise, accurate, and only respond to direct queries.

Task: Write a SQL statement to count employees hired after 2020.
Response:

```
The model then fills in the SQL code.

### 3. Long Instructional Prompt
Some prompts contain detailed behavioral instructions, such as:
> “If you don’t know the answer, respond with ‘I don’t know.’ Do not speculate.”

Such long prompts became famous with Bing Chat and academic studies analyzing complex prompt designs.

## Chain-of-Thought Prompting

**Chain-of-thought (CoT) prompting** (introduced in 2022) helps models solve multi-step problems by asking them to explain intermediate reasoning steps.

Example: Solving a word problem  
Instead of directly answering, the model outputs intermediate steps like:
1. Rewriting the problem
2. Setting up an equation
3. Solving step-by-step

This improves performance on complex reasoning tasks.

### Why It Works
Two main hypotheses:
1. **Training Data Hypothesis:** The model has seen similar step-by-step examples during pre-training.
2. **Decomposition Hypothesis:** Breaking problems into smaller parts makes them easier for the model to handle.

---

## Least-to-Most Prompting

**Least-to-most prompting** builds on CoT prompting.  
Here, the model first solves **simpler subproblems**, then uses those results to solve **more complex problems**.

Example task:
> Given a list of words, concatenate the last letter of each.

The model starts with a single word, gets its last letter, then progressively handles longer lists—using each previous solution as a building block.

This method outperforms standard chain-of-thought prompting.

---

## Concept-Based Prompting

In research from **DeepMind**, models were asked to first **state relevant concepts or equations** (from physics or chemistry) before solving problems.  

When LLMs first reasoned about *why* or *how* to solve equations, their accuracy increased significantly.
