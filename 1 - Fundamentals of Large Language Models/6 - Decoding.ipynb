# Large Language Models — Decoding

Hello. In the previous lessons, we've discussed how large language models compute distributions over vocabulary words and how we can affect those distributions. In this lesson, we'll talk about how we can take these distributions and **generate text**, a process known as **decoding**.

---

## The Decoding Process

Let’s revisit our example:  
> “I wrote to the zoo to send me a pet. They sent me a—”

The large language model (LLM) produces a **distribution over vocabulary words**, representing the probabilities of what the next word could be. The question is — how do we turn this distribution into a sequence of words?

Decoding happens **one word at a time** in an **iterative process**:
1. The model is given some input text.  
2. It produces a probability distribution over possible next words.  
3. A word is selected (based on a chosen decoding strategy).  
4. The selected word is appended to the input.  
5. The process repeats until an **End of Sequence (EOS)** token is produced.

The model does **not** generate an entire sentence or document in one go — each token is chosen step by step.

---

## Greedy Decoding

The simplest decoding method is **greedy decoding**.  
In this strategy, we pick the **word with the highest probability** at each step.

Example:  
In our sentence, the word **“dog”** has the highest probability, so the model selects “dog,” appends it, and re-evaluates the next probabilities.  
When the EOS token receives the highest probability, generation stops.

Result:
> “I wrote to the zoo to send me a pet. They sent me a dog.”

Greedy decoding is straightforward but deterministic — it always produces the same output for the same input.

---

## Sampling-Based Decoding

To introduce **diversity** in the generated text, we can use **non-deterministic** strategies — that is, methods that include **random sampling**.

Instead of always choosing the top word, we **sample** from the distribution:
- Sometimes from all words, weighted by probability.  
- Sometimes only from the top few high-probability words.

Example:  
If we sample randomly, the model might pick **“small”**, append it, then sample again and pick **“red”**, which increases the probability of **“panda.”**  
The final sentence could be:
> “I wrote to the zoo to send me a pet. They sent me a small red panda.”

This randomness produces more **creative** and **varied** text.

---

## Temperature in Decoding

An important parameter in sampling-based decoding is **temperature**, which controls how “peaked” or “flat” the probability distribution is.

- **Low temperature (< 1):**  
  - Makes the distribution sharper.  
  - The highest-probability words become even more likely.  
  - The model behaves more deterministically (similar to greedy decoding).

- **High temperature (> 1):**  
  - Flattens the distribution.  
  - Makes low-probability (rare) words more likely.  
  - Increases creativity and variability.

Temperature affects the **spread** of probabilities but not their **ranking** — the most likely word remains the most likely regardless of temperature.

So, when temperature is low → predictable output.  
When temperature is high → diverse, imaginative output.

---

## When to Use Each Method

- **Low temperature / Greedy decoding:**  
  Ideal for factual, precise tasks like answering direct questions.

- **High temperature / Sampling:**  
  Useful for open-ended tasks like storytelling, dialogue, or brainstorming — where creativity and diversity are valued.

Decoding strategy selection depends on whether you want **accuracy** or **originality** in the generated text.

---

## Other Common Decoding Methods

### 1. Greedy Decoding  
Selects the most probable word at each step. Simple but can lack diversity.

### 2. Nucleus Sampling (Top-p Sampling)  
Samples words only from the smallest subset of the vocabulary whose cumulative probability exceeds a threshold *p* (e.g., 0.9).  
This balances creativity and coherence.

### 3. Beam Search  
Generates **multiple possible sequences simultaneously**, keeping only the most probable ones as generation continues.  
Beam search is **not greedy**, yet it often yields outputs with **higher overall probability** than greedy decoding.

---

## Summary

Decoding is how LLMs turn probability distributions into text — word by word.  
Different strategies trade off between **determinism** and **diversity**:
- Greedy decoding → predictable but limited.
- Sampling → varied and creative.
- Beam search and nucleus sampling → smart compromises between the two.

Decoding remains an active research area, with many methods designed to shape the characteristics of generated text.
