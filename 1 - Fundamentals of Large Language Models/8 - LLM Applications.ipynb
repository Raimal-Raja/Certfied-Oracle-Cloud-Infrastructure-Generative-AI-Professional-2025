## 🔍 1. Retrieval-Augmented Generation (RAG)

**Definition:**  
Retrieval-Augmented Generation (RAG) combines **retrieval** and **generation** to improve factual accuracy and reduce hallucination.

### 💡 How RAG Works
1. **User Input**: The user provides a query or question.  
2. **Retrieval Step**:  
   - The system converts the question into a **search query**.  
   - This query retrieves relevant documents from a database or corpus.  
3. **Generation Step**:  
   - The **LLM** is given both the **user question** and the **retrieved documents**.  
   - The LLM uses this combined input to generate a more accurate answer.

### 🧩 Benefits
- **Reduces hallucination** by grounding responses in factual data.  
- **Improves accuracy** for tasks like Q&A, dialogue, and fact-checking.  
- **Non-parametric improvement:**  
  You don’t need to retrain the model — just **add more documents** to the corpus.

### ⚙️ Example Application
- **Customer support:**  
  Connect the LLM to your company’s manuals or documentation.  
  When users ask questions, the model retrieves relevant sections and answers directly.

## 💻 2. LLMs for Code (Code Models)

**Definition:**  
These are LLMs trained specifically on **code, comments, and technical documentation**.

### 🧠 Examples
- **GitHub Copilot**  
- **OpenAI Codex**  
- **Code Llama**

### 🪄 Capabilities
- **Code completion**  
- **Documentation generation**  
- **Function writing** from a natural-language description  

For example:
> “Write a Python function that checks if a number is prime.”

The model can generate the full function automatically.

### ⚙️ Advantages
- Reduces the need for **boilerplate code**  
- Helps with **unfamiliar programming languages**  
- Useful for **syntax lookups** and **quick prototypes**

### ⚠️ Limitations
- Struggles with complex or multi-step bugs  
- Studies show models can **automatically patch real bugs** < 15% of the time

## 🖼️ 3. Multimodal Models

**Definition:**  
Models trained on **multiple data modalities** — such as text, images, and audio.

### 🎨 Capabilities
- Generate images or videos from text descriptions  
- Perform cross-modal tasks (e.g., “describe this image”)  
- Power applications like **image captioning** and **text-to-image synthesis**

### 🌀 Diffusion Models
Unlike LLMs (which generate text **one token at a time**), diffusion models generate **images all at once**.

#### Process:
1. Start with pure **noise**.  
2. Iteratively **refine all pixels simultaneously** until a coherent image appears.  

These are the foundation of tools like **DALL·E** and **Stable Diffusion**.

#### Why Joint Decoding for Text is Hard
- **Text length is variable** — we don’t know how many words to generate.  
- **Words are discrete**, unlike continuous pixel values.  
- Continuous refinement doesn’t naturally apply to discrete language tokens.

## 🤖 4. Language Agents

**Definition:**  
LLMs used as **autonomous agents** for **sequential decision-making** tasks.

### 🧩 Examples
- Playing chess  
- Operating software autonomously  
- Browsing the web to complete a task  

### ⚙️ How They Work
1. The model operates within an **environment**.  
2. It performs **actions** (e.g., searching or clicking).  
3. The environment **responds** (e.g., search results).  
4. The model observes, reasons, and **chooses the next action**.  
5. This loop continues until the model achieves its goal.

### 🧭 Example Framework: ReAct
- Prompts the model to produce **“thoughts”** — internal reasoning steps:
  - What is my goal?  
  - What have I done so far?  
  - What should I do next?  

This creates a structured chain-of-thought process.

## 🧰 5. Tool Use by LLMs

LLMs can learn to **use tools and APIs** to expand their capabilities.

### 🧠 Example:
Instead of performing arithmetic via text generation:
1. The LLM says: “I will use a calculator.”  
2. It generates an **API call** to compute the result.  
3. The system feeds the API’s response back into the model.  

This enables:
- External computation  
- Access to up-to-date information  
- Integration with specialized systems (databases, search engines, etc.)

## 🧩 6. Reasoning and Planning

Recent research focuses on **training LLMs to reason** more effectively.

### Why It Matters
- Enables **long-term planning**  
- Allows agents to handle **unfamiliar environments**  
- Supports **high-level strategy** and **goal decomposition**

### Analogy
Just like humans, reasoning helps LLMs:
- Adapt to new challenges  
- Make decisions over multiple steps  
- Improve performance in complex, dynamic tasks
