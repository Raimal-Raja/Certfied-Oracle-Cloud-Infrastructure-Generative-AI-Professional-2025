primally used in QA, Where the model has access to (retrieved) support documents for a query

claimed to reduce hallucination
multi-document QA via fancy decoding RAG-tok

Idea has gotten a lot of traction

used in dialogue, QA fact-checking, slot filling, entity-linking
non-parametric; in theory, the same model can answer questions about any corpus

this system is conceptually very simple

## code models
insteadd of training on written language, train on code and comments

co-pilot, codex, code llama

complete partly written functions, synthesize programs from docstrings, debugging

Largely successful: > 85% of people using co-pilot feel more productive

great fit between training data (code + comments) and test-time task (write code + comments). also code is structured easier to learn 

this is unlike LLM, which are trained on a wide varity of internet text and used for many purposes (other  than generating internet text) code models have (arguably) narrower  scope
