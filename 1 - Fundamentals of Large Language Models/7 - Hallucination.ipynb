# Large Language Models — Hallucination

Welcome back. Last time, we spoke about various methods of decoding. In this lesson, we'll talk about **hallucination** — one of the most challenging issues in large language model (LLM) behavior.

---

## What Is Hallucination?

Hallucination can be defined in several ways, but for our purposes, it refers to **text generated by a model that is not grounded in any data** the model has been exposed to.  
In other words, hallucination occurs when an LLM produces statements that are **unsupported**, **nonsensical**, or **factually incorrect**, based on either its training data or input context.

For example, consider the statement:  
> “In the United States, people gradually adopted the practice of driving on the left side of the road.”  

This statement is **fluent** and grammatically correct, but **factually wrong** — people in the U.S. drive on the right side. Such output would be considered a **hallucination**.

---

## The Subtle Nature of Hallucinations

While some hallucinations are blatant, many are **subtle and hard to detect**.  
A model might introduce a small factual inaccuracy, such as:  
> “Barack Obama was the first president of the United States.”  

This sentence sounds natural but is obviously incorrect.  
Such subtle errors can be dangerous, particularly when users lack the background knowledge to verify the information. This makes hallucination a **major challenge for safely deploying LLMs**.

A useful way to think about this comes from Professor Samir Singh of UC Irvine, who said:  
> “Think about LLMs as chameleons. They're trying to generate text that blends in with human-generated text — whether or not it’s true. It just needs to sound true.”

Some even argue that **all LLM-generated text is hallucinated**, in the sense that it is always *generated*, not *recalled*, and just happens to be correct much of the time.

---

## The Risks of Hallucination

Hallucination is problematic because:
- Generated text is often **fluent and persuasive**, making falsehoods hard to spot.  
- Errors can be **minor or severe**, affecting credibility and safety.  
- There is currently **no reliable way** to completely eliminate hallucinations.  

Despite their sophistication, LLMs remain **probabilistic text generators**, not fact-verification systems.

---

## Mitigation and Best Practices

While hallucination cannot be eliminated entirely, researchers and developers are exploring strategies to **reduce** it.

### 1. Retrieval-Augmented Generation (RAG)
Evidence suggests that **retrieval-augmented systems** hallucinate less than pure, zero-shot models.  
RAG combines an LLM with an external knowledge base or document retrieval system, grounding the output in real-world data.

### 2. Measuring Groundedness
Another growing area of research involves **measuring how well LLM outputs are supported** by evidence.  
This is done through **Natural Language Inference (NLI)** — a task where a model checks whether a given document (the *premise*) supports a generated sentence (the *hypothesis*).

Some NLI-based models, such as **TRUE**, can automatically assess whether an LLM’s statement is grounded in retrieved or reference documents.  
These systems work reasonably well but tend to be **conservative**, erring on the side of caution.

### 3. Grounded Question Answering
In this approach, the model not only answers questions but also **cites its sources**.  
This increases transparency and helps users verify where information comes from.

### 4. Attribution and Citation
Ongoing work focuses on **attributing** generated text to verifiable sources, improving accountability and reliability in LLM outputs.

---

## Conclusion

Hallucination remains one of the biggest open challenges in the field of large language models.  
While models often generate correct, coherent, and fluent text, they can — and do — produce statements that are **unfounded, misleading, or false**.

The research community is actively developing methods to **detect**, **measure**, and **reduce** hallucination through grounded generation, citation, and verification systems.
