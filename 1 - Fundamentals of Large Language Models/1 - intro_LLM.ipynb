## What is a Language Model?

At its core, a **language model** is a **probabilistic model of text**.  
This means it predicts the **likelihood** of the next word in a sequence based on the words that came before it.

Let’s consider an example:

> “I wrote to the zoo to send me a pet. They sent me a —”

This is a line from a well-known children’s book.  
A **language model** takes the preceding words and computes a **distribution of probabilities** over all possible next words in its **vocabulary**.  

- The model “knows” a set of words — its **vocabulary**.  
- For each word in that vocabulary, it assigns a **probability** that the word should appear next.  
- It will only output words that exist in its vocabulary — **no new words** outside of it.

Thus, a language model’s job is to generate or predict text based on **statistical likelihood**.

## What Makes a Model “Large”?

So what does the “large” in **Large Language Model (LLM)** really mean?

Technically, the **L** in LLM stands for **large**, but this term doesn’t have a fixed definition.  
It generally refers to the **number of parameters** in the model — parameters being the internal variables the model learns during training.

However, there’s **no agreed-upon threshold** for when a model becomes “large” or “extra large.”  
For example:
- Some researchers might call **GPT-4** or **PaLM** “large” because they contain hundreds of billions of parameters.  
- Others might even refer to smaller models like **BERT** as “large” because of their architectural complexity.

So, while “large” is somewhat subjective, in practice, **LLMs** generally refer to the **modern transformer-based models** capable of generating or understanding text at scale.

## What Can LLMs Do?

At a fundamental level, an **LLM**:
- Accepts a sequence of text as **input** (called a *prompt*).  
- Produces a **probability distribution** over its vocabulary for what might come next.  

The **distribution values** — the numbers you might see under words on a visualization — determine how **likely** each word is to be selected as the next token.

But this raises important questions:
1. Can we **influence** or **control** this distribution?  
2. What **mechanisms** allow us to guide the model’s output?  
3. How do we go from probability distributions to **actual generated text**?

These are the motivating questions for the rest of this module.

### 2. Affecting the Distribution
We’ll then explore **two ways** to influence a model’s probability distribution:

- **Prompting:**  
  This method **does not change** the model’s internal parameters.  
  Instead, it uses creative input phrasing to shape output behavior.

- **Training:**  
  This method **does** change the model’s parameters.  
  It involves fine-tuning or retraining the model on new datasets to adjust its internal understanding.

---

### 3. Decoding
The final major topic is **decoding** — the process of **generating text** from an LLM.  
Decoding is how we turn probability distributions into coherent, human-like sentences, paragraphs, or documents.

We’ll explore several decoding strategies that balance creativity and coherence in generated text.
