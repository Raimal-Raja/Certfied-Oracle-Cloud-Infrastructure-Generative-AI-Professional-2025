# Lesson 2: Large Language Model Architectures

## Introduction
Hello, and welcome to **Lesson 2** of this OCI learning module on **Large Language Models (LLMs)**.  
In this lesson, we’ll explore **LLM architectures**, focusing on the two most common types:

1. **Encoders**  
2. **Decoders**

These architectures correspond to two different model capabilities:
- **Embedding** – representing text numerically (for meaning or similarity)
- **Text Generation** – producing text outputs word-by-word or token-by-token

Before diving into each, it’s important to note that **both encoders and decoders** are built upon a shared foundation — the **Transformer** architecture.

---

## The Transformer Architecture
The **Transformer** was introduced in the groundbreaking 2017 paper *“Attention Is All You Need.”*  
This architecture completely transformed the fields of **Natural Language Processing (NLP)** and **Machine Learning**.

While we won’t go deep into its internal workings here, remember that **encoders** and **decoders** both rely on **self-attention** mechanisms — a way for models to understand relationships between words, regardless of their position in the sentence.

For a deeper dive, there are excellent tutorials and visualizations available online.

---

## Model Size and Parameters
When we refer to the **size** of a model, we’re talking about the **number of trainable parameters** — the learned weights that define the model’s knowledge and behavior.

To illustrate this, imagine a taxonomy of model types:
- **Encoders** (e.g., BERT)
- **Decoders** (e.g., GPT-4, LLaMA, Cohere Command)
- **Encoder-Decoder Hybrids** (e.g., T5, BART)

Two observations:
1. The **y-axis** of the model taxonomy increases **by orders of magnitude** (each step is a 10× jump, not linear).  
2. **Decoder models** tend to be much larger than encoders.

This is largely due to historical reasons — smaller encoders work fine for embeddings, while **larger models** are needed for **high-quality text generation**.  
However, **recent research** shows progress in making smaller models capable of fluent text generation through improved efficiency and architecture design.

---

## Understanding Encoders
**Encoders** are designed to **embed text** — that is, to convert words or sentences into **vector representations**.

For example, consider the sentence:

> “They sent me a …”

An **encoder** (like BERT) transforms each word in this sentence into a **vector**, and it also produces a **sentence-level embedding** representing the entire input.

These embeddings capture **semantic meaning** — words and phrases with similar meanings will have **similar vector representations**.

Encoders are commonly used for:
- **Classification** (e.g., sentiment analysis)
- **Clustering**
- **Regression**
- **Semantic or vector search**

### Example: Semantic Search
Imagine you want to find similar documents to a user’s query:
1. Encode each document in your database into a vector.
2. Encode the input query into a vector.
3. Compute the **similarity** (e.g., cosine similarity) between the query and document vectors.
4. Return the most similar documents.

This is the basis for **semantic search** and **recommendation systems**.

## Understanding Decoders
Now, let’s move to **decoder models** — these are designed for **text generation**.

Popular decoder models include:
- **Cohere Command**
- **GPT-4**
- **LLaMA**

### How Decoders Work
Decoders take an **input sequence of tokens** and predict the **next token** in that sequence based on a probability distribution over the vocabulary.

Key point:  
A **decoder only produces one token at a time.**

To generate longer text:
1. Feed the initial input into the model to predict the next token.  
2. Append that token to the input sequence.  
3. Feed the updated sequence back into the model.  
4. Repeat until the desired text length is reached.

This iterative process is what allows models like **ChatGPT** to generate coherent multi-sentence responses.

However, this process is **computationally expensive**, especially for very large models.

### Use Cases
- **Text generation**
- **Dialogue systems**
- **Question answering**
- **Creative writing**

Typically, decoders are **not used for embeddings** — that’s the job of encoders.

---

## Encoder-Decoder Models
The final category is **Encoder-Decoder Models**, which combine the strengths of both architectures.  
They’re particularly suited for **sequence-to-sequence tasks** like **translation**.

### How It Works
1. The **encoder** processes the input text (e.g., English sentence) and produces embeddings for each token and the whole sentence.  
2. These embeddings are passed to the **decoder**, which generates output text (e.g., in French) one token at a time.  
3. Each new token generated is fed back into the decoder to predict the next word.

This structure enables the model to map one sequence (input) to another (output).

Examples include:
- **T5 (Text-to-Text Transfer Transformer)**
- **BART**
- **MarianMT**

## Summary Table: Architectures and Tasks

| **Task Type**             | **Encoder** | **Decoder** | **Encoder-Decoder** |
|----------------------------|-------------|--------------|----------------------|
| Text Embedding             | ✅ Yes      | ❌ No        | ✅ Sometimes         |
| Text Generation            | ❌ No       | ✅ Yes       | ✅ Yes               |
| Classification / Regression| ✅ Yes      | ✅ Sometimes | ✅ Yes               |
| Machine Translation        | ❌ No       | ✅ Sometimes | ✅ Yes               |
| Semantic Search            | ✅ Yes      | ❌ No        | ✅ Sometimes         |

**Note:**  
This table isn’t rigid — any model *could* technically perform any task, but these pairings are what’s most common and effective in practice.

---

## Key Takeaways
- **Encoders** create **numeric representations** (embeddings) of text for tasks like search or classification.  
- **Decoders** generate **text** by predicting one token at a time.  
- **Encoder-Decoders** combine both and are powerful for **sequence-to-sequence tasks** such as translation.  
- **Model size** varies drastically, often measured by the number of parameters (millions to trillions).  
- **Transformers** are the foundational architecture behind all modern LLMs.
