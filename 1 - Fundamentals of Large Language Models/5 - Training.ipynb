# Large Language Models — Training

Welcome back to the OCI learning module on large language models. In the previous two lessons, we talked about **prompting** as a way to affect the model's distribution over vocabulary words. But prompting is not the only option for affecting this distribution. In fact, a more significant way to alter the distribution over vocabulary is **training**, which we'll talk about in this lesson.

Remember that prompting is, in effect, simply changing the **input** to an LLM. The process is highly sensitive. Small changes in the prompt can yield large changes in the distribution over words. Moreover, since the model's parameters are fixed, by using prompting alone, we're limited in how much we can change the model's word distribution.

In this way, prompting is sometimes insufficient. For example, in **domain adaptation**, when a model trained on one domain is used in an entirely new one, something more substantial is needed. Unlike prompting, **training** actually changes the parameters of the model.

At a high level, training is the process of giving the model an input, having it guess a corresponding output — for example, completing a sentence or answering a question — and then altering the parameters so that next time, it generates something closer to the correct answer.

Once we alter the model’s parameters, the distribution over words changes — ideally for the better. There are multiple ways to train or modify the model’s parameters, each with different costs and benefits.

---

## Approaches to Training

### 1. Fine-tuning
In **fine-tuning**, we take a **pre-trained model** (e.g., BERT) and a **labeled dataset** for a specific task, and train the model by altering all of its parameters.  
This was the standard approach around **2019**.  
While training a model like BERT was considered expensive then, it’s much cheaper than training today’s massive models with hundreds of billions of parameters.

Because full fine-tuning is costly, we’ve developed cheaper alternatives, such as **parameter-efficient fine-tuning** methods.

---

### 2. Parameter-Efficient Fine-Tuning
In these methods, only a **small subset** of the model's parameters are trained, or a few new ones are added.

A popular method is **LoRA** (*Low-Rank Adaptation*), where the model’s main parameters remain fixed, and small additional parameters are trained.  
This approach drastically reduces computational cost while maintaining strong performance.

---

### 3. Soft Prompting
**Soft prompting** is another low-cost training method, but conceptually different from LoRA.

Here, parameters are **added to the prompt** rather than the model. These are specialized “virtual words” learned during training to cue the model for specific tasks.

Unlike manual prompting, a soft prompt is **learned automatically** — the parameters representing these artificial tokens are randomly initialized and then fine-tuned through training.

---

### 4. Continual Pretraining
**Continual pretraining** changes all model parameters (like fine-tuning) but does **not require labeled data**.

Instead of predicting specific labels, the model is simply trained to **predict the next word** in large amounts of new text data.  
This method is particularly effective for adapting models to new domains — for example, taking a general model and retraining it on millions of scientific articles to specialize it in that domain.

---

## Cost of Training

Training costs vary widely depending on:
- Model size (number of parameters)
- Amount of training data
- Number and type of GPUs
- Duration of training

A chart illustrating this would show:
- **Right side:** Text generation — relatively cheap, done on 1 GPU in seconds (for 7B models). Larger models (150B+) may need 8–16 GPUs.
- **Middle:** Parameter-efficient training — moderate cost, requiring a few GPUs for several hours.
- **Left side:** Full pretraining — extremely expensive, needing **hundreds or thousands of GPUs** running for days.

---

## The “Cramming” Study

A recent research paper introduced **cramming**, exploring how far one can train a model with **just one GPU and 24 hours**.  
The findings demonstrate the limits of efficient training and serve as a thought-provoking look at how to maximize results under tight resource constraints.
